```python
df = df[::-1]
df_0 = df[['ゲーム', '台 番', '日期']]
df_0.columns = ['game', 'no', 'date']
df_0 = df_0.reset_index(drop=True)


def check_condition(group):
    condition = (group['game'].shift(-1) < 100) & \
                (group['game'].shift(-2) < 100) & \
                (group['game'].shift(-3) < 100)
    
    same_group = (group['no'] == group['no'].shift(-1)) & \
                 (group['no'] == group['no'].shift(-2)) & \
                 (group['no'] == group['no'].shift(-3)) & \
                 (group['date'] == group['date'].shift(-1)) & \
                 (group['date'] == group['date'].shift(-2)) & \
                 (group['date'] == group['date'].shift(-3))
    
    return condition & same_group


df_0['target'] = df_0.groupby(['no', 'date']).apply(check_condition).reset_index(level=[0,1], drop=True)
# df_0['target'] = df_0['target'].astype(int)
df_0['target_1'] = df_0['target'].shift(1)
df_0[:40]
# df_0['target'] = df_0['target'].astype(int)
# df_0['target_1'] = df_0['target_1'].astype(int)
# df_0[(df_0['target'] == True )&(df_0['target'] == False )]
df_0 = df_0.dropna()

df_0['target'] = df_0['target'].apply(lambda x:int(x))
df_0['target_1'] = df_0['target_1'].apply(lambda x:int(x))
```

```python
df_1  = df_0[(df_0['target']==1)&(df_0['target_1']==0)]
df_1['target_0'] = 1
df_1
```

```python
df_2 = df_0[['game','no','date']].join(df_1[['game','no','date','target_0']].set_index(['game','no','date']),on=['game','no','date'],how='left')
```

```python
df_2['target_0'] = df_2['target_0'].fillna(0)
```
