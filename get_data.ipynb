```python
from selenium import webdriver
from selenium.webdriver.firefox.service import Service
from selenium.webdriver.firefox.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.action_chains import ActionChains
import time
import random 

def get_coordinates_and_click(driver, locator, timeout=10):
    wait = WebDriverWait(driver, timeout)
    # 等待元素出现并定位它
    element = wait.until(EC.presence_of_element_located(locator))
    
    location = element.location
    size = element.size
    
    center_x = location['x'] + size['width'] / 2
    center_y = location['y'] + size['height'] / 2
        
    # 使用ActionChains点击元素
    actions = ActionChains(driver)
    actions.move_to_element(element).click().perform()
    
    return center_x, center_y

def input_text(driver, locator, text, timeout=10):
    wait = WebDriverWait(driver, timeout)
    
    input_element = wait.until(EC.presence_of_element_located(locator))
    
    actions = ActionChains(driver)
    actions.move_to_element(input_element).click().perform()
    
    for char in text:
        input_element.send_keys(char)
        time.sleep(random.uniform(0.1, 0.5))  # 随机延迟，模拟人工输入
    

print("设置 Firefox 选项")
# 设置 Firefox 选项
firefox_options = Options()
# firefox_options.add_argument("--headless")  

# 设置 WebDriver
service = Service("/Users/lalu/Downloads/geckodriver")  # 替换为你的geckodriver路径
driver = webdriver.Firefox(service=service, options=firefox_options)

print("调整窗口大小")
driver.set_window_size(1443, 986)  # 设置窗口大小为1920x1080
# 设置等待
wait = WebDriverWait(driver, random.uniform(1,4))  # 1-4秒随机等待

print("打开网站")
# 打开网站
driver.get("https://www.d-deltanet.com/pc/MypageLoginTop.do?redirectLogin=0&skskb=")
print("网站已打开")

print("尝试定位并点击'同意'按钮")
# 使用函数获取坐标并点击按钮
x, y = get_coordinates_and_click(driver, (By.CSS_SELECTOR, "div.agree button"))

print(f"等待 {5} 秒")
time.sleep(random.uniform(1,4))

print("定位并输入邮箱地址")
input_text(driver, (By.ID, "mypage_email"), "roakira0825@outlook.com")
time.sleep(random.uniform(1,4))

print("定位并输入密码")
input_text(driver, (By.ID, "mypage_pass"), "Asfire0825")

print(f"等待 {5} 秒")
time.sleep(random.uniform(1,4))

print("尝试执行登录操作")
login_script = "cmdButton_push(document.MypageLoginActionForm);"
driver.execute_script(login_script)
print("登录操作已执行")

print(f"等待 {5} 秒")
time.sleep(random.uniform(1,4))
print("等待完成")
# 前面的导入和设置保持不变

print("等待页面加载完成")
WebDriverWait(driver, 20).until(EC.presence_of_element_located((By.TAG_NAME, "body")))

print("页面标题:", driver.title)
print("当前 URL:", driver.current_url)



print("尝试定位并点击'スーパーＤ’ステーション錦糸町店'链接")

# 使用正确的撇号字符，并添加更精确的定位策略
locator_strategies = [
    (By.CSS_SELECTOR, "li#myhall_10019053 a"),  # 直接使用元素的ID和标签
    (By.XPATH, "//li[@id='myhall_10019053']//a"),
    (By.LINK_TEXT, "スーパーＤ’ステーション錦糸町店"),  # 使用正确的撇号字符
    (By.PARTIAL_LINK_TEXT, "スーパーＤ’ステーション錦糸町店"),
    (By.XPATH, "//a[contains(text(), \"スーパーＤ’ステーション錦糸町店\")]")
]

wait = WebDriverWait(driver, 20)  # 20秒等待时间

# 遍历定位策略
for locator in locator_strategies:
    print(f"尝试使用定位策略: {locator}")
    try:
        element = wait.until(EC.element_to_be_clickable(locator))
        print("元素找到，尝试点击")
        element.click()
        print(f"已点击'スーパーＤ’ステーション錦糸町店'链接")
        break
    except Exception as e:
        print(f"使用当前定位策略未找到元素，错误信息：{e}")
else:
    print("所有定位策略都失败了")
    print("页面源代码:")
    print(driver.page_source)
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import random
import math

# 假设 driver 已经定义并指向正确的页面

# 创建等待对象
wait = WebDriverWait(driver, 20)

# 等待表格出现
table = wait.until(EC.presence_of_element_located((By.XPATH, "//table[@class='slot']")))
print("找到表格元素")

# 查找所有商品名称元素
name_elements = table.find_elements(By.XPATH, ".//span")
total_elements = len(name_elements)
print(f"总共有 {total_elements} 个商品名称")

# 初始化商品名称列表
product_names = []

# 遍历所有的商品名称元素
for idx, name_elem in enumerate(name_elements):
    product_name = name_elem.text.strip()
    product_names.append(product_name)
    print(f"{idx + 1}. {product_name}")
    
    # 将元素滚动到视口中
    driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", name_elem)
    time.sleep(random.uniform(0.3,0.9))  # 等待页面滚动完成
    
    # 模拟鼠标移动到商品名称上
    actions = ActionChains(driver)
    actions.move_to_element(name_elem).perform()
    # 添加延迟，模拟人工浏览
    time.sleep(random.uniform(0.5, 1.5))

print("\n所有商品名称已提取并存储到列表中。")
```

```python
from selenium.webdriver.common.action_chains import ActionChains
import time
import random

def click_button_by_text(driver, button_text):
    # 定位到指定文本的链接
    link = driver.find_element(By.XPATH, f"//a[contains(text(),'{button_text}')]")
    
    # 获取链接的位置信息
    location = link.location
    size = link.size
    center_x = location['x'] + size['width'] / 2
    center_y = location['y'] + size['height'] / 2

    # 输出坐标信息
    print(f"'{button_text}' 按钮的坐标：({center_x}, {center_y})")

    # 使用 ActionChains 来模拟鼠标点击
    actions = ActionChains(driver)
    actions.move_to_element_with_offset(link, size['width'] / 2, size['height'] / 2)
    actions.click()
    actions.perform()

    # 点击后的处理（如等待页面加载）
    time.sleep(random.uniform(1, 3))

import pandas as pd

def extract_table_data(driver, table_xpath, timeout=10):
    # 使用显式等待确保表格完全加载
    wait = WebDriverWait(driver, timeout)
    wait.until(EC.presence_of_element_located((By.XPATH, table_xpath)))
    # 获取表格的所有行
    rows = driver.find_elements(By.XPATH, f"{table_xpath}/tbody/tr")

    # 初始化一个列表来存放每行的数据
    data = []

    # 遍历所有行
    for row in rows:
        # 获取行内的所有单元格
        cols = row.find_elements(By.TAG_NAME, 'td')
        # 提取每个单元格的文本，如果单元格内包含span标签，则提取span的文本
        row_data = []
        for col in cols:
            if col.text:
                row_data.append(col.text)
            else:
                span = col.find_element(By.TAG_NAME, 'span')
                row_data.append(span.text if span else '--')
        data.append(row_data)

    # 列标题，这里需要确保列的数量和标题对应正确
    columns = ['大当り回数', '種類', '時間', 'ゲーム', '獲得数']

    # 创建DataFrame
    df = pd.DataFrame(data, columns=columns)

    return df

# click_button_by_text(driver, "5前")
# time.sleep(random.uniform(0.5,3))
# df = extract_table_data(driver, "//table")
# time.sleep(random.uniform(0.5,3))
```

```python
from selenium.webdriver.common.action_chains import ActionChains

def click_link_by_text_and_return(driver, link_text):
    """点击指定文本链接并通过ActionChains模拟鼠标点击，确保链接被点击."""
    link = driver.find_element(By.XPATH, f"//a[contains(text(), '{link_text}')]")
    ActionChains(driver).move_to_element(link).click(link).perform()
    print(f"Clicked on link: {link_text}")
    time.sleep(random.uniform(1, 3))  # 等待页面加载

# 示例用法
click_link_by_text_and_return(driver, "大当り一覧")
```
